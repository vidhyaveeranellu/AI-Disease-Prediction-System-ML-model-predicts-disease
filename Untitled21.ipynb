```python
# ============================================================
# COMPLETE PYTORCH CODE
# ACRIMA GLAUCOMA CLASSIFICATION
# DATASET LABELS ARE CREATED FROM "_g_" IN THE FILENAME
# ============================================================

# Install packages
!pip install -q torch torchvision scikit-learn matplotlib seaborn pillow pandas

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader

from torchvision import transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ============================================================
# 2. MOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")

# ============================================================
# 3. CONFIGURATION
# ============================================================

# Your exact dataset path
DATA_DIR = "/content/drive/MyDrive/Database/Database/Images"

# Image size specified in the base paper
IMAGE_SIZE = 200

# Batch size
BATCH_SIZE = 16

# Change this to 100 for the second experiment
EPOCHS = 50

# Learning rate specified in the base paper
LEARNING_RATE = 0.0001

# Dropout value
DROPOUT_RATE = 0.5

# Base-paper filter configuration
FILTERS = [16, 32, 64]

# The paper mentions threshold = 0.9.
# Use 0.5 for standard binary classification.
# Change to 0.9 for the paper threshold experiment.
THRESHOLD = 0.5

# Validation ratio
VALIDATION_RATIO = 0.05

# Random seed
SEED = 42

# Use GPU if available
DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)
print("Dataset path:", DATA_DIR)
print("Dataset path exists:", os.path.exists(DATA_DIR))

if not os.path.exists(DATA_DIR):
    raise FileNotFoundError(
        f"Dataset folder not found: {DATA_DIR}"
    )

# ============================================================
# 4. REPRODUCIBILITY
# ============================================================

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


set_seed(SEED)

# ============================================================
# 5. CREATE RESULTS FOLDER
# ============================================================

RESULTS_DIR = "/content/drive/MyDrive/ACRIMA_CNN_Results"

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

# ============================================================
# 6. LOAD DATASET USING FILENAME LABELS
# ============================================================

valid_extensions = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
)

data = []

for filename in sorted(os.listdir(DATA_DIR)):

    if filename.lower().endswith(valid_extensions):

        filepath = os.path.join(
            DATA_DIR,
            filename
        )

        # ACRIMA filename rule:
        # "_g_" means glaucoma
        # otherwise normal
        if "_g_" in filename.lower():
            label = 1
            class_name = "glaucoma"
        else:
            label = 0
            class_name = "normal"

        data.append({
            "filepath": filepath,
            "label": label,
            "class_name": class_name,
            "filename": filename
        })

if len(data) == 0:
    raise ValueError(
        "No image files were found in DATA_DIR."
    )

dataframe = pd.DataFrame(data)

print("\nDataset summary:")
print(dataframe["class_name"].value_counts())

print("\nTotal images:", len(dataframe))
print("Glaucoma images:", int((dataframe["label"] == 1).sum()))
print("Normal images:", int((dataframe["label"] == 0).sum()))

# Verify that both classes exist
if dataframe["label"].nunique() < 2:
    raise ValueError(
        "Only one class was detected. Check the filename rule '_g_'."
    )

# ============================================================
# 7. TRAIN-VALIDATION SPLIT
# ============================================================

train_dataframe, validation_dataframe = train_test_split(
    dataframe,
    test_size=VALIDATION_RATIO,
    random_state=SEED,
    stratify=dataframe["label"]
)

train_dataframe = train_dataframe.reset_index(
    drop=True
)

validation_dataframe = validation_dataframe.reset_index(
    drop=True
)

print("\nTraining images:", len(train_dataframe))
print("Validation images:", len(validation_dataframe))

print("\nTraining class distribution:")
print(train_dataframe["class_name"].value_counts())

print("\nValidation class distribution:")
print(validation_dataframe["class_name"].value_counts())

# ============================================================
# 8. IMAGE TRANSFORMATIONS
# ============================================================

train_transforms = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.RandomAffine(
        degrees=0,
        translate=(0.05, 0.05),
        scale=(0.95, 1.05)
    ),

    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

validation_transforms = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# ============================================================
# 9. CUSTOM PYTORCH DATASET
# ============================================================

class ACRIMADataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):
        self.dataframe = dataframe.reset_index(
            drop=True
        )

        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):

        image_path = self.dataframe.loc[
            index,
            "filepath"
        ]

        label = self.dataframe.loc[
            index,
            "label"
        ]

        image = Image.open(
            image_path
        ).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        label = torch.tensor(
            label,
            dtype=torch.float32
        )

        return image, label


train_dataset = ACRIMADataset(
    dataframe=train_dataframe,
    transform=train_transforms
)

validation_dataset = ACRIMADataset(
    dataframe=validation_dataframe,
    transform=validation_transforms
)

# ============================================================
# 10. DATA LOADERS
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

# ============================================================
# 11. BASE-PAPER CNN MODEL
# ============================================================

class BasePaperCNN(nn.Module):

    def __init__(
        self,
        filters=(16, 32, 64),
        dropout_rate=0.5
    ):
        super().__init__()

        self.features = nn.Sequential(

            # First convolution block
            nn.Conv2d(
                in_channels=3,
                out_channels=filters[0],
                kernel_size=3,
                padding=1
            ),

            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            # Second convolution block
            nn.Conv2d(
                in_channels=filters[0],
                out_channels=filters[1],
                kernel_size=3,
                padding=1
            ),

            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            # Third convolution block
            nn.Conv2d(
                in_channels=filters[1],
                out_channels=filters[2],
                kernel_size=3,
                padding=1
            ),

            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            nn.Dropout(
                p=dropout_rate
            )
        )

        # Image dimension:
        # 200 -> 100 -> 50 -> 25
        flattened_features = (
            filters[2] * 25 * 25
        )

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(
                flattened_features,
                128
            ),

            nn.ReLU(inplace=True),

            nn.Dropout(
                p=dropout_rate
            ),

            # Binary output logit
            nn.Linear(
                128,
                1
            )
        )

    def forward(self, x):

        x = self.features(x)
        x = self.classifier(x)

        return x


model = BasePaperCNN(
    filters=FILTERS,
    dropout_rate=DROPOUT_RATE
).to(DEVICE)

print("\nModel:")
print(model)

# ============================================================
# 12. LOSS FUNCTION AND OPTIMIZER
# ============================================================

criterion = nn.BCEWithLogitsLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

# ============================================================
# 13. TRAINING FUNCTION
# ============================================================

def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device
):
    model.train()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    for images, labels in loader:

        images = images.to(device)

        labels = labels.to(device)
        labels = labels.view(-1, 1)

        optimizer.zero_grad()

        logits = model(images)

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        optimizer.step()

        total_loss += (
            loss.item() * images.size(0)
        )

        probabilities = torch.sigmoid(
            logits
        )

        predictions = (
            probabilities >= 0.5
        ).float()

        total_correct += (
            predictions == labels
        ).sum().item()

        total_samples += labels.size(0)

    average_loss = (
        total_loss / total_samples
    )

    accuracy = (
        total_correct / total_samples
    )

    return average_loss, accuracy

# ============================================================
# 14. VALIDATION FUNCTION
# ============================================================

def validate_model(
    model,
    loader,
    criterion,
    device
):
    model.eval()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    all_labels = []
    all_probabilities = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)

            labels = labels.to(device)
            labels = labels.view(-1, 1)

            logits = model(images)

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item() * images.size(0)
            )

            probabilities = torch.sigmoid(
                logits
            )

            predictions = (
                probabilities >= 0.5
            ).float()

            total_correct += (
                predictions == labels
            ).sum().item()

            total_samples += labels.size(0)

            all_labels.extend(
                labels.cpu().numpy().ravel()
            )

            all_probabilities.extend(
                probabilities.cpu().numpy().ravel()
            )

    average_loss = (
        total_loss / total_samples
    )

    accuracy = (
        total_correct / total_samples
    )

    return (
        average_loss,
        accuracy,
        np.array(all_labels).astype(int),
        np.array(all_probabilities)
    )

# ============================================================
# 15. TRAIN THE MODEL
# ============================================================

history = {
    "train_loss": [],
    "train_accuracy": [],
    "validation_loss": [],
    "validation_accuracy": []
}

best_validation_loss = float("inf")

best_model_path = os.path.join(
    RESULTS_DIR,
    "base_paper_cnn_best.pth"
)

print("\nStarting training...\n")

for epoch in range(EPOCHS):

    train_loss, train_accuracy = train_one_epoch(
        model=model,
        loader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=DEVICE
    )

    validation_loss, validation_accuracy, _, _ = validate_model(
        model=model,
        loader=validation_loader,
        criterion=criterion,
        device=DEVICE
    )

    history["train_loss"].append(
        train_loss
    )

    history["train_accuracy"].append(
        train_accuracy
    )

    history["validation_loss"].append(
        validation_loss
    )

    history["validation_accuracy"].append(
        validation_accuracy
    )

    print(
        f"Epoch [{epoch + 1:03d}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.4f} | "
        f"Validation Loss: {validation_loss:.4f} | "
        f"Validation Accuracy: {validation_accuracy:.4f}"
    )

    if validation_loss < best_validation_loss:

        best_validation_loss = validation_loss

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "validation_loss": validation_loss,
                "filters": FILTERS,
                "image_size": IMAGE_SIZE,
                "threshold": THRESHOLD
            },
            best_model_path
        )

print("\nTraining completed.")

# ============================================================
# 16. LOAD THE BEST MODEL
# ============================================================

checkpoint = torch.load(
    best_model_path,
    map_location=DEVICE
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

# ============================================================
# 17. FINAL PREDICTIONS
# ============================================================

all_true_labels = []
all_predicted_probabilities = []

with torch.no_grad():

    for images, labels in validation_loader:

        images = images.to(DEVICE)

        logits = model(images)

        probabilities = torch.sigmoid(
            logits
        )

        all_true_labels.extend(
            labels.numpy().ravel()
        )

        all_predicted_probabilities.extend(
            probabilities.cpu().numpy().ravel()
        )

all_true_labels = np.array(
    all_true_labels
).astype(int)

all_predicted_probabilities = np.array(
    all_predicted_probabilities
)

all_predicted_labels = (
    all_predicted_probabilities >= THRESHOLD
).astype(int)

# ============================================================
# 18. METRICS
# ============================================================

cm = confusion_matrix(
    all_true_labels,
    all_predicted_labels,
    labels=[0, 1]
)

tn, fp, fn, tp = cm.ravel()

accuracy = accuracy_score(
    all_true_labels,
    all_predicted_labels
)

precision = precision_score(
    all_true_labels,
    all_predicted_labels,
    zero_division=0
)

sensitivity = recall_score(
    all_true_labels,
    all_predicted_labels,
    zero_division=0
)

specificity = (
    tn / (tn + fp)
    if (tn + fp) > 0
    else 0.0
)

f1 = f1_score(
    all_true_labels,
    all_predicted_labels,
    zero_division=0
)

try:
    auc = roc_auc_score(
        all_true_labels,
        all_predicted_probabilities
    )
except ValueError:
    auc = float("nan")

print("\nFinal validation results")
print("==============================")
print("Label mapping:")
print("0 = normal")
print("1 = glaucoma")
print("Threshold:", THRESHOLD)
print("Filters:", FILTERS)
print("Epochs:", EPOCHS)
print()
print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
print()
print(f"Accuracy:    {accuracy:.4f}")
print(f"Sensitivity: {sensitivity:.4f}")
print(f"Specificity: {specificity:.4f}")
print(f"Precision:   {precision:.4f}")
print(f"F1-score:    {f1:.4f}")
print(f"ROC-AUC:     {auc:.4f}")

print("\nClassification report:")
print(
    classification_report(
        all_true_labels,
        all_predicted_labels,
        target_names=[
            "normal",
            "glaucoma"
        ],
        zero_division=0
    )
)

# ============================================================
# 19. SAVE METRICS
# ============================================================

metrics_file = os.path.join(
    RESULTS_DIR,
    "metrics.txt"
)

with open(metrics_file, "w") as file:

    file.write(
        "Base-Paper PyTorch CNN Results\n"
    )

    file.write(
        "================================\n"
    )

    file.write(
        f"Dataset path: {DATA_DIR}\n"
    )

    file.write(
        f"Total images: {len(dataframe)}\n"
    )

    file.write(
        f"Training images: {len(train_dataframe)}\n"
    )

    file.write(
        f"Validation images: {len(validation_dataframe)}\n"
    )

    file.write(
        f"Filters: {FILTERS}\n"
    )

    file.write(
        f"Epochs: {EPOCHS}\n"
    )

    file.write(
        f"Threshold: {THRESHOLD}\n"
    )

    file.write(
        f"TP: {tp}\n"
    )

    file.write(
        f"TN: {tn}\n"
    )

    file.write(
        f"FP: {fp}\n"
    )

    file.write(
        f"FN: {fn}\n"
    )

    file.write(
        f"Accuracy: {accuracy:.4f}\n"
    )

    file.write(
        f"Sensitivity: {sensitivity:.4f}\n"
    )

    file.write(
        f"Specificity: {specificity:.4f}\n"
    )

    file.write(
        f"Precision: {precision:.4f}\n"
    )

    file.write(
        f"F1-score: {f1:.4f}\n"
    )

    file.write(
        f"ROC-AUC: {auc:.4f}\n"
    )

# ============================================================
# 20. PLOT CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(6, 5)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "normal",
        "glaucoma"
    ],
    yticklabels=[
        "normal",
        "glaucoma"
    ]
)

plt.xlabel("Predicted class")
plt.ylabel("True class")
plt.title("Base-Paper CNN Confusion Matrix")

plt.tight_layout()

confusion_matrix_path = os.path.join(
    RESULTS_DIR,
    "confusion_matrix.png"
)

plt.savefig(
    confusion_matrix_path,
    dpi=300
)

plt.show()

# ============================================================
# 21. PLOT TRAINING HISTORY
# ============================================================

epoch_numbers = range(
    1,
    EPOCHS + 1
)

plt.figure(
    figsize=(12, 5)
)

plt.subplot(
    1,
    2,
    1
)

plt.plot(
    epoch_numbers,
    history["train_accuracy"],
    label="Training accuracy"
)

plt.plot(
    epoch_numbers,
    history["validation_accuracy"],
    label="Validation accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy Curve")
plt.legend()

plt.subplot(
    1,
    2,
    2
)

plt.plot(
    epoch_numbers,
    history["train_loss"],
    label="Training loss"
)

plt.plot(
    epoch_numbers,
    history["validation_loss"],
    label="Validation loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Curve")
plt.legend()

plt.tight_layout()

history_path = os.path.join(
    RESULTS_DIR,
    "training_history.png"
)

plt.savefig(
    history_path,
    dpi=300
)

plt.show()

# ============================================================
# 22. SAVE FINAL MODEL
# ============================================================

final_model_path = os.path.join(
    RESULTS_DIR,
    "base_paper_cnn_final.pth"
)

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "filters": FILTERS,
        "image_size": IMAGE_SIZE,
        "threshold": THRESHOLD
    },
    final_model_path
)

print("\nCompleted successfully.")
print("Best model:", best_model_path)
print("Final model:", final_model_path)
print("Metrics:", metrics_file)
print("Results folder:", RESULTS_DIR)
```

```python
# ============================================================
# COMPLETE PYTORCH CODE
# ACRIMA GLAUCOMA CLASSIFICATION
# DATASET LABELS ARE CREATED FROM "_g_" IN THE FILENAME
# ============================================================

# Install packages
!pip install -q torch torchvision scikit-learn matplotlib seaborn pillow pandas

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader

from torchvision import transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ============================================================
# 2. MOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")

# ============================================================
# 3. CONFIGURATION
# ============================================================

# Your exact dataset path
DATA_DIR = "/content/drive/MyDrive/Database/Database/Images"

# Image size specified in the base paper
IMAGE_SIZE = 200

# Batch size
BATCH_SIZE = 16

# Change this to 100 for the second experiment
EPOCHS = 100

# Learning rate specified in the base paper
LEARNING_RATE = 0.0001

# Dropout value
DROPOUT_RATE = 0.5

# Base-paper filter configuration
FILTERS = [32, 64, 128]

# The paper mentions threshold = 0.9.
# Use 0.5 for standard binary classification.
# Change to 0.9 for the paper threshold experiment.
THRESHOLD = 0.5

# Validation ratio
VALIDATION_RATIO = 0.05

# Random seed
SEED = 42

# Use GPU if available
DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)
print("Dataset path:", DATA_DIR)
print("Dataset path exists:", os.path.exists(DATA_DIR))

if not os.path.exists(DATA_DIR):
    raise FileNotFoundError(
        f"Dataset folder not found: {DATA_DIR}"
    )

# ============================================================
# 4. REPRODUCIBILITY
# ============================================================

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


set_seed(SEED)

# ============================================================
# 5. CREATE RESULTS FOLDER
# ============================================================

RESULTS_DIR = "/content/drive/MyDrive/ACRIMA_CNN_Results"

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

# ============================================================
# 6. LOAD DATASET USING FILENAME LABELS
# ============================================================

valid_extensions = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
)

data = []

for filename in sorted(os.listdir(DATA_DIR)):

    if filename.lower().endswith(valid_extensions):

        filepath = os.path.join(
            DATA_DIR,
            filename
        )

        # ACRIMA filename rule:
        # "_g_" means glaucoma
        # otherwise normal
        if "_g_" in filename.lower():
            label = 1
            class_name = "glaucoma"
        else:
            label = 0
            class_name = "normal"

        data.append({
            "filepath": filepath,
            "label": label,
            "class_name": class_name,
            "filename": filename
        })

if len(data) == 0:
    raise ValueError(
        "No image files were found in DATA_DIR."
    )

dataframe = pd.DataFrame(data)

print("\nDataset summary:")
print(dataframe["class_name"].value_counts())

print("\nTotal images:", len(dataframe))
print("Glaucoma images:", int((dataframe["label"] == 1).sum()))
print("Normal images:", int((dataframe["label"] == 0).sum()))

# Verify that both classes exist
if dataframe["label"].nunique() < 2:
    raise ValueError(
        "Only one class was detected. Check the filename rule '_g_'."
    )

# ============================================================
# 7. TRAIN-VALIDATION SPLIT
# ============================================================

train_dataframe, validation_dataframe = train_test_split(
    dataframe,
    test_size=VALIDATION_RATIO,
    random_state=SEED,
    stratify=dataframe["label"]
)

train_dataframe = train_dataframe.reset_index(
    drop=True
)

validation_dataframe = validation_dataframe.reset_index(
    drop=True
)

print("\nTraining images:", len(train_dataframe))
print("Validation images:", len(validation_dataframe))

print("\nTraining class distribution:")
print(train_dataframe["class_name"].value_counts())

print("\nValidation class distribution:")
print(validation_dataframe["class_name"].value_counts())

# ============================================================
# 8. IMAGE TRANSFORMATIONS
# ============================================================

train_transforms = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.RandomAffine(
        degrees=0,
        translate=(0.05, 0.05),
        scale=(0.95, 1.05)
    ),

    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

validation_transforms = transforms.Compose([
    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# ============================================================
# 9. CUSTOM PYTORCH DATASET
# ============================================================

class ACRIMADataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):
        self.dataframe = dataframe.reset_index(
            drop=True
        )

        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):

        image_path = self.dataframe.loc[
            index,
            "filepath"
        ]

        label = self.dataframe.loc[
            index,
            "label"
        ]

        image = Image.open(
            image_path
        ).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        label = torch.tensor(
            label,
            dtype=torch.float32
        )

        return image, label


train_dataset = ACRIMADataset(
    dataframe=train_dataframe,
    transform=train_transforms
)

validation_dataset = ACRIMADataset(
    dataframe=validation_dataframe,
    transform=validation_transforms
)

# ============================================================
# 10. DATA LOADERS
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

# ============================================================
# 11. BASE-PAPER CNN MODEL
# ============================================================

class BasePaperCNN(nn.Module):

    def __init__(
        self,
        filters=(16, 32, 64),
        dropout_rate=0.5
    ):
        super().__init__()

        self.features = nn.Sequential(

            # First convolution block
            nn.Conv2d(
                in_channels=3,
                out_channels=filters[0],
                kernel_size=3,
                padding=1
            ),

            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            # Second convolution block
            nn.Conv2d(
                in_channels=filters[0],
                out_channels=filters[1],
                kernel_size=3,
                padding=1
            ),

            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            # Third convolution block
            nn.Conv2d(
                in_channels=filters[1],
                out_channels=filters[2],
                kernel_size=3,
                padding=1
            ),

            nn.ReLU(inplace=True),

            nn.MaxPool2d(
                kernel_size=2,
                stride=2
            ),

            nn.Dropout(
                p=dropout_rate
            )
        )

        # Image dimension:
        # 200 -> 100 -> 50 -> 25
        flattened_features = (
            filters[2] * 25 * 25
        )

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(
                flattened_features,
                128
            ),

            nn.ReLU(inplace=True),

            nn.Dropout(
                p=dropout_rate
            ),

            # Binary output logit
            nn.Linear(
                128,
                1
            )
        )

    def forward(self, x):

        x = self.features(x)
        x = self.classifier(x)

        return x


model = BasePaperCNN(
    filters=FILTERS,
    dropout_rate=DROPOUT_RATE
).to(DEVICE)

print("\nModel:")
print(model)

# ============================================================
# 12. LOSS FUNCTION AND OPTIMIZER
# ============================================================

criterion = nn.BCEWithLogitsLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

# ============================================================
# 13. TRAINING FUNCTION
# ============================================================

def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device
):
    model.train()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    for images, labels in loader:

        images = images.to(device)

        labels = labels.to(device)
        labels = labels.view(-1, 1)

        optimizer.zero_grad()

        logits = model(images)

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        optimizer.step()

        total_loss += (
            loss.item() * images.size(0)
        )

        probabilities = torch.sigmoid(
            logits
        )

        predictions = (
            probabilities >= 0.5
        ).float()

        total_correct += (
            predictions == labels
        ).sum().item()

        total_samples += labels.size(0)

    average_loss = (
        total_loss / total_samples
    )

    accuracy = (
        total_correct / total_samples
    )

    return average_loss, accuracy

# ============================================================
# 14. VALIDATION FUNCTION
# ============================================================

def validate_model(
    model,
    loader,
    criterion,
    device
):
    model.eval()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    all_labels = []
    all_probabilities = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)

            labels = labels.to(device)
            labels = labels.view(-1, 1)

            logits = model(images)

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item() * images.size(0)
            )

            probabilities = torch.sigmoid(
                logits
            )

            predictions = (
                probabilities >= 0.5
            ).float()

            total_correct += (
                predictions == labels
            ).sum().item()

            total_samples += labels.size(0)

            all_labels.extend(
                labels.cpu().numpy().ravel()
            )

            all_probabilities.extend(
                probabilities.cpu().numpy().ravel()
            )

    average_loss = (
        total_loss / total_samples
    )

    accuracy = (
        total_correct / total_samples
    )

    return (
        average_loss,
        accuracy,
        np.array(all_labels).astype(int),
        np.array(all_probabilities)
    )

# ============================================================
# 15. TRAIN THE MODEL
# ============================================================

history = {
    "train_loss": [],
    "train_accuracy": [],
    "validation_loss": [],
    "validation_accuracy": []
}

best_validation_loss = float("inf")

best_model_path = os.path.join(
    RESULTS_DIR,
    "base_paper_cnn_best.pth"
)

print("\nStarting training...\n")

for epoch in range(EPOCHS):

    train_loss, train_accuracy = train_one_epoch(
        model=model,
        loader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=DEVICE
    )

    validation_loss, validation_accuracy, _, _ = validate_model(
        model=model,
        loader=validation_loader,
        criterion=criterion,
        device=DEVICE
    )

    history["train_loss"].append(
        train_loss
    )

    history["train_accuracy"].append(
        train_accuracy
    )

    history["validation_loss"].append(
        validation_loss
    )

    history["validation_accuracy"].append(
        validation_accuracy
    )

    print(
        f"Epoch [{epoch + 1:03d}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.4f} | "
        f"Validation Loss: {validation_loss:.4f} | "
        f"Validation Accuracy: {validation_accuracy:.4f}"
    )

    if validation_loss < best_validation_loss:

        best_validation_loss = validation_loss

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "validation_loss": validation_loss,
                "filters": FILTERS,
                "image_size": IMAGE_SIZE,
                "threshold": THRESHOLD
            },
            best_model_path
        )

print("\nTraining completed.")

# ============================================================
# 16. LOAD THE BEST MODEL
# ============================================================

checkpoint = torch.load(
    best_model_path,
    map_location=DEVICE
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

# ============================================================
# 17. FINAL PREDICTIONS
# ============================================================

all_true_labels = []
all_predicted_probabilities = []

with torch.no_grad():

    for images, labels in validation_loader:

        images = images.to(DEVICE)

        logits = model(images)

        probabilities = torch.sigmoid(
            logits
        )

        all_true_labels.extend(
            labels.numpy().ravel()
        )

        all_predicted_probabilities.extend(
            probabilities.cpu().numpy().ravel()
        )

all_true_labels = np.array(
    all_true_labels
).astype(int)

all_predicted_probabilities = np.array(
    all_predicted_probabilities
)

all_predicted_labels = (
    all_predicted_probabilities >= THRESHOLD
).astype(int)

# ============================================================
# 18. METRICS
# ============================================================

cm = confusion_matrix(
    all_true_labels,
    all_predicted_labels,
    labels=[0, 1]
)

tn, fp, fn, tp = cm.ravel()

accuracy = accuracy_score(
    all_true_labels,
    all_predicted_labels
)

precision = precision_score(
    all_true_labels,
    all_predicted_labels,
    zero_division=0
)

sensitivity = recall_score(
    all_true_labels,
    all_predicted_labels,
    zero_division=0
)

specificity = (
    tn / (tn + fp)
    if (tn + fp) > 0
    else 0.0
)

f1 = f1_score(
    all_true_labels,
    all_predicted_labels,
    zero_division=0
)

try:
    auc = roc_auc_score(
        all_true_labels,
        all_predicted_probabilities
    )
except ValueError:
    auc = float("nan")

print("\nFinal validation results")
print("==============================")
print("Label mapping:")
print("0 = normal")
print("1 = glaucoma")
print("Threshold:", THRESHOLD)
print("Filters:", FILTERS)
print("Epochs:", EPOCHS)
print()
print("TP:", tp)
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
print()
print(f"Accuracy:    {accuracy:.4f}")
print(f"Sensitivity: {sensitivity:.4f}")
print(f"Specificity: {specificity:.4f}")
print(f"Precision:   {precision:.4f}")
print(f"F1-score:    {f1:.4f}")
print(f"ROC-AUC:     {auc:.4f}")

print("\nClassification report:")
print(
    classification_report(
        all_true_labels,
        all_predicted_labels,
        target_names=[
            "normal",
            "glaucoma"
        ],
        zero_division=0
    )
)

# ============================================================
# 19. SAVE METRICS
# ============================================================

metrics_file = os.path.join(
    RESULTS_DIR,
    "metrics.txt"
)

with open(metrics_file, "w") as file:

    file.write(
        "Base-Paper PyTorch CNN Results\n"
    )

    file.write(
        "================================\n"
    )

    file.write(
        f"Dataset path: {DATA_DIR}\n"
    )

    file.write(
        f"Total images: {len(dataframe)}\n"
    )

    file.write(
        f"Training images: {len(train_dataframe)}\n"
    )

    file.write(
        f"Validation images: {len(validation_dataframe)}\n"
    )

    file.write(
        f"Filters: {FILTERS}\n"
    )

    file.write(
        f"Epochs: {EPOCHS}\n"
    )

    file.write(
        f"Threshold: {THRESHOLD}\n"
    )

    file.write(
        f"TP: {tp}\n"
    )

    file.write(
        f"TN: {tn}\n"
    )

    file.write(
        f"FP: {fp}\n"
    )

    file.write(
        f"FN: {fn}\n"
    )

    file.write(
        f"Accuracy: {accuracy:.4f}\n"
    )

    file.write(
        f"Sensitivity: {sensitivity:.4f}\n"
    )

    file.write(
        f"Specificity: {specificity:.4f}\n"
    )

    file.write(
        f"Precision: {precision:.4f}\n"
    )

    file.write(
        f"F1-score: {f1:.4f}\n"
    )

    file.write(
        f"ROC-AUC: {auc:.4f}\n"
    )

# ============================================================
# 20. PLOT CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(6, 5)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "normal",
        "glaucoma"
    ],
    yticklabels=[
        "normal",
        "glaucoma"
    ]
)

plt.xlabel("Predicted class")
plt.ylabel("True class")
plt.title("Base-Paper CNN Confusion Matrix")

plt.tight_layout()

confusion_matrix_path = os.path.join(
    RESULTS_DIR,
    "confusion_matrix.png"
)

plt.savefig(
    confusion_matrix_path,
    dpi=300
)

plt.show()

# ============================================================
# 21. PLOT TRAINING HISTORY
# ============================================================

epoch_numbers = range(
    1,
    EPOCHS + 1
)

plt.figure(
    figsize=(12, 5)
)

plt.subplot(
    1,
    2,
    1
)

plt.plot(
    epoch_numbers,
    history["train_accuracy"],
    label="Training accuracy"
)

plt.plot(
    epoch_numbers,
    history["validation_accuracy"],
    label="Validation accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy Curve")
plt.legend()

plt.subplot(
    1,
    2,
    2
)

plt.plot(
    epoch_numbers,
    history["train_loss"],
    label="Training loss"
)

plt.plot(
    epoch_numbers,
    history["validation_loss"],
    label="Validation loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Curve")
plt.legend()

plt.tight_layout()

history_path = os.path.join(
    RESULTS_DIR,
    "training_history.png"
)

plt.savefig(
    history_path,
    dpi=300
)

plt.show()

# ============================================================
# 22. SAVE FINAL MODEL
# ============================================================

final_model_path = os.path.join(
    RESULTS_DIR,
    "base_paper_cnn_final.pth"
)

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "filters": FILTERS,
        "image_size": IMAGE_SIZE,
        "threshold": THRESHOLD
    },
    final_model_path
)

print("\nCompleted successfully.")
print("Best model:", best_model_path)
print("Final model:", final_model_path)
print("Metrics:", metrics_file)
print("Results folder:", RESULTS_DIR)
```

```python
# ============================================================
# RELIABILITY-AWARE EXPLAINABLE GLaucoma CNN
# MC DROPOUT + IMAGE QUALITY + GRAD-CAM
# ============================================================

import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn

from PIL import Image
from torchvision import transforms

# ============================================================
# 1. Configuration
# ============================================================

MC_PASSES = 30

# Final decision thresholds
GLAUCOMA_THRESHOLD = 0.90
NORMAL_THRESHOLD = 0.10

# Uncertainty threshold will be calibrated automatically
# from validation images below.
UNCERTAINTY_PERCENTILE = 90

# Quality threshold
QUALITY_THRESHOLD = 0.60

# Approximate optic-disc region.
# This is a prototype because ACRIMA optic-disc masks are not
# available in the current dataset.
DISC_CENTER_X = 0.50
DISC_CENTER_Y = 0.50
DISC_RADIUS = 0.22

RELIABILITY_RESULTS_DIR = os.path.join(
    RESULTS_DIR,
    "reliability_results"
)

os.makedirs(
    RELIABILITY_RESULTS_DIR,
    exist_ok=True
)

print(
    "Reliability results will be saved to:",
    RELIABILITY_RESULTS_DIR
)

# ============================================================
# 2. Verify model and variables
# ============================================================

required_variables = [
    "model",
    "DEVICE",
    "validation_transforms",
    "validation_dataframe",
    "RESULTS_DIR"
]

missing_variables = [
    variable
    for variable in required_variables
    if variable not in globals()
]

if missing_variables:
    raise RuntimeError(
        "Run the base CNN training cell first. "
        f"Missing variables: {missing_variables}"
    )

model = model.to(DEVICE)

# ============================================================
# 3. Image-quality functions
# ============================================================

def normalize_score(value, minimum, maximum):
    return float(
        np.clip(
            (value - minimum) /
            (maximum - minimum + 1e-8),
            0.0,
            1.0
        )
    )


def calculate_image_quality(image):
    """
    Returns a prototype image-quality score from 0 to 1.

    Components:
    - Sharpness using Laplacian variance
    - Brightness
    - Contrast
    - Border coverage
    """

    image_array = np.asarray(image).astype(np.uint8)

    gray = cv2.cvtColor(
        image_array,
        cv2.COLOR_RGB2GRAY
    )

    # Sharpness: higher Laplacian variance generally
    # means a sharper image.
    sharpness = cv2.Laplacian(
        gray,
        cv2.CV_64F
    ).var()

    # Brightness normalized around an acceptable range
    brightness = float(
        np.mean(gray) / 255.0
    )

    # Contrast normalized using grayscale standard deviation
    contrast = float(
        np.std(gray) / 128.0
    )

    # Estimate black-border percentage
    black_pixels = np.mean(
        gray < 10
    )

    sharpness_score = normalize_score(
        sharpness,
        minimum=20,
        maximum=500
    )

    # Brightness around 0.45-0.65 is considered more useful
    brightness_score = float(
        np.clip(
            1.0 - abs(brightness - 0.55) / 0.55,
            0.0,
            1.0
        )
    )

    contrast_score = normalize_score(
        contrast,
        minimum=0.10,
        maximum=0.80
    )

    border_score = float(
        np.clip(
            1.0 - black_pixels / 0.70,
            0.0,
            1.0
        )
    )

    quality_score = (
        0.35 * sharpness_score +
        0.25 * brightness_score +
        0.25 * contrast_score +
        0.15 * border_score
    )

    return {
        "quality_score": float(
            np.clip(quality_score, 0.0, 1.0)
        ),
        "sharpness": sharpness,
        "brightness": brightness,
        "contrast": contrast,
        "black_border_fraction": float(black_pixels),
        "sharpness_score": sharpness_score,
        "brightness_score": brightness_score,
        "contrast_score": contrast_score,
        "border_score": border_score
    }

# ============================================================
# 4. Enable MC Dropout
# ============================================================

def enable_dropout_layers(model):
    """
    Keep only dropout layers in train mode while the rest
    of the model remains in evaluation mode.
    """

    model.eval()

    for module in model.modules():

        if isinstance(
            module,
            (
                nn.Dropout,
                nn.Dropout2d,
                nn.Dropout3d
            )
        ):
            module.train()


# ============================================================
# 5. MC Dropout prediction
# ============================================================

def mc_dropout_prediction(
    image_tensor,
    model,
    device,
    passes=30
):
    """
    Returns:
    - mean glaucoma probability
    - predictive variance
    - predictive entropy
    - all stochastic probabilities
    """

    enable_dropout_layers(model)

    probabilities = []

    with torch.no_grad():

        for _ in range(passes):

            logits = model(
                image_tensor.to(device)
            )

            probability = torch.sigmoid(
                logits
            ).item()

            probabilities.append(
                probability
            )

    probabilities = np.array(
        probabilities
    )

    mean_probability = float(
        np.mean(probabilities)
    )

    variance = float(
        np.var(probabilities)
    )

    entropy_probability = np.clip(
        mean_probability,
        1e-7,
        1 - 1e-7
    )

    entropy = -(
        entropy_probability *
        np.log(entropy_probability)
        +
        (1 - entropy_probability) *
        np.log(1 - entropy_probability)
    )

    model.eval()

    return {
        "mean_probability": mean_probability,
        "variance": variance,
        "entropy": float(entropy),
        "probabilities": probabilities
    }

# ============================================================
# 6. Grad-CAM implementation
# ============================================================

class GradCAM:

    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer

        self.activations = None
        self.gradients = None

        self.forward_handle = (
            target_layer.register_forward_hook(
                self.save_activations
            )
        )

        self.backward_handle = (
            target_layer.register_full_backward_hook(
                self.save_gradients
            )
        )

    def save_activations(
        self,
        module,
        inputs,
        output
    ):
        self.activations = output

    def save_gradients(
        self,
        module,
        grad_input,
        grad_output
    ):
        self.gradients = grad_output[0]

    def generate(self, image_tensor):

        self.model.eval()
        self.model.zero_grad()

        output = self.model(
            image_tensor
        )

        # Explain glaucoma score
        target_score = output[:, 0].sum()

        target_score.backward()

        gradients = self.gradients
        activations = self.activations

        weights = torch.mean(
            gradients,
            dim=(2, 3),
            keepdim=True
        )

        cam = torch.sum(
            weights * activations,
            dim=1
        )

        cam = torch.relu(cam)

        cam = cam[0].detach().cpu().numpy()

        cam = cv2.resize(
            cam,
            (IMAGE_SIZE, IMAGE_SIZE)
        )

        cam_min = cam.min()
        cam_max = cam.max()

        cam = (
            cam - cam_min
        ) / (
            cam_max - cam_min + 1e-8
        )

        return output.detach(), cam

    def close(self):

        self.forward_handle.remove()
        self.backward_handle.remove()

# ============================================================
# 7. Approximate optic-disc mask
# ============================================================

def create_approximate_disc_mask(
    image_height,
    image_width,
    center_x=0.50,
    center_y=0.50,
    radius=0.22
):
    """
    Prototype optic-disc mask.

    Replace this with a real optic-disc detector or
    annotation mask for the final research version.
    """

    y, x = np.ogrid[
        :image_height,
        :image_width
    ]

    center_pixel_x = center_x * image_width
    center_pixel_y = center_y * image_height

    radius_pixels = radius * min(
        image_height,
        image_width
    )

    mask = (
        (x - center_pixel_x) ** 2 +
        (y - center_pixel_y) ** 2
        <= radius_pixels ** 2
    )

    return mask.astype(np.float32)

# ============================================================
# 8. Grad-CAM attention alignment
# ============================================================

def calculate_attention_alignment(
    heatmap,
    disc_mask,
    activation_threshold=0.60
):
    """
    Calculates the proportion of high Grad-CAM activation
    inside the approximate optic-disc region.
    """

    high_activation = (
        heatmap >= activation_threshold
    ).astype(np.float32)

    total_activation = np.sum(
        high_activation
    )

    if total_activation == 0:
        return 0.0

    activation_inside_disc = np.sum(
        high_activation * disc_mask
    )

    alignment = (
        activation_inside_disc /
        total_activation
    )

    return float(
        np.clip(alignment, 0.0, 1.0)
    )

# ============================================================
# 9. Combined reliability score
# ============================================================

def calculate_reliability_score(
    quality_score,
    uncertainty,
    attention_alignment
):
    """
    Reliability score:
    high quality + low uncertainty + high
    attention alignment = high reliability.
    """

    # Normalize uncertainty approximately to 0-1.
    normalized_uncertainty = np.clip(
        uncertainty / 0.25,
        0.0,
        1.0
    )

    reliability = (
        0.40 * quality_score +
        0.35 * (1.0 - normalized_uncertainty) +
        0.25 * attention_alignment
    )

    return float(
        np.clip(reliability, 0.0, 1.0)
    )

# ============================================================
# 10. Decision rule
# ============================================================

def final_decision(
    glaucoma_probability,
    uncertainty,
    quality_score,
    attention_alignment,
    uncertainty_threshold,
    quality_threshold=0.60
):
    reliability_score = calculate_reliability_score(
        quality_score=quality_score,
        uncertainty=uncertainty,
        attention_alignment=attention_alignment
    )

    if quality_score < quality_threshold:
        decision = (
            "review required: poor image quality"
        )

    elif uncertainty > uncertainty_threshold:
        decision = (
            "review required: uncertain prediction"
        )

    elif attention_alignment < 0.10:
        decision = (
            "review required: weak optic-disc alignment"
        )

    elif glaucoma_probability >= GLAUCOMA_THRESHOLD:
        decision = "glaucoma"

    elif glaucoma_probability <= NORMAL_THRESHOLD:
        decision = "normal"

    else:
        decision = (
            "review required: borderline probability"
        )

    return decision, reliability_score

# ============================================================
# 11. Choose Grad-CAM target layer
# ============================================================

# Your BasePaperCNN has:
# features[0] = Conv2d(3, 16, ...)
# features[3] = Conv2d(16, 32, ...)
# features[6] = Conv2d(32, 64, ...)
#
# The last convolutional layer is features[6].
target_layer = model.features[6]

gradcam = GradCAM(
    model=model,
    target_layer=target_layer
)

# ============================================================
# 12. Calibrate uncertainty threshold
# ============================================================

print("\nCalibrating uncertainty threshold...")

calibration_uncertainties = []

for index in range(
    len(validation_dataframe)
):

    row = validation_dataframe.iloc[index]

    image = Image.open(
        row["filepath"]
    ).convert("RGB")

    image_tensor = validation_transforms(
        image
    ).unsqueeze(0)

    mc_result = mc_dropout_prediction(
        image_tensor=image_tensor,
        model=model,
        device=DEVICE,
        passes=MC_PASSES
    )

    calibration_uncertainties.append(
        mc_result["variance"]
    )

uncertainty_threshold = float(
    np.percentile(
        calibration_uncertainties,
        UNCERTAINTY_PERCENTILE
    )
)

print(
    "Calibrated uncertainty threshold:",
    uncertainty_threshold
)

# ============================================================
# 13. Analyze one image
# ============================================================

def analyze_one_image(
    image_path,
    save_output=True
):
    if not os.path.exists(image_path):
        raise FileNotFoundError(
            f"Image not found: {image_path}"
        )

    original_image = Image.open(
        image_path
    ).convert("RGB")

    image_array = np.asarray(
        original_image
    )

    image_tensor = validation_transforms(
        original_image
    ).unsqueeze(0)

    # Image quality
    quality_result = calculate_image_quality(
        original_image
    )

    # MC Dropout
    mc_result = mc_dropout_prediction(
        image_tensor=image_tensor,
        model=model,
        device=DEVICE,
        passes=MC_PASSES
    )

    glaucoma_probability = (
        mc_result["mean_probability"]
    )

    uncertainty = (
        mc_result["variance"]
    )

    # Grad-CAM
    cam_output, heatmap = gradcam.generate(
        image_tensor.to(DEVICE)
    )

    # Approximate optic-disc mask
    disc_mask = create_approximate_disc_mask(
        image_height=IMAGE_SIZE,
        image_width=IMAGE_SIZE,
        center_x=DISC_CENTER_X,
        center_y=DISC_CENTER_Y,
        radius=DISC_RADIUS
    )

    attention_alignment = (
        calculate_attention_alignment(
            heatmap=heatmap,
            disc_mask=disc_mask
        )
    )

    decision, reliability_score = final_decision(
        glaucoma_probability=glaucoma_probability,
        uncertainty=uncertainty,
        quality_score=quality_result["quality_score"],
        attention_alignment=attention_alignment,
        uncertainty_threshold=uncertainty_threshold,
        quality_threshold=QUALITY_THRESHOLD
    )

    # Print result
    print("\nPrediction report")
    print("==============================")
    print("Image:", os.path.basename(image_path))
    print(
        f"Glaucoma probability: "
        f"{glaucoma_probability:.4f}"
    )
    print(
        f"Normal probability: "
        f"{1 - glaucoma_probability:.4f}"
    )
    print(
        f"Uncertainty variance: "
        f"{uncertainty:.6f}"
    )
    print(
        f"Image quality score: "
        f"{quality_result['quality_score']:.4f}"
    )
    print(
        f"Attention alignment: "
        f"{attention_alignment:.4f}"
    )
    print(
        f"Reliability score: "
        f"{reliability_score:.4f}"
    )
    print("Final decision:", decision)

    # Prepare display image
    display_image = cv2.resize(
        image_array,
        (IMAGE_SIZE, IMAGE_SIZE)
    )

    heatmap_color = cv2.applyColorMap(
        np.uint8(255 * heatmap),
        cv2.COLORMAP_JET
    )

    heatmap_color = cv2.cvtColor(
        heatmap_color,
        cv2.COLOR_BGR2RGB
    )

    overlay = cv2.addWeighted(
        display_image,
        0.55,
        heatmap_color,
        0.45,
        0
    )

    disc_outline = display_image.copy()

    disc_center = (
        int(DISC_CENTER_X * IMAGE_SIZE),
        int(DISC_CENTER_Y * IMAGE_SIZE)
    )

    disc_radius = int(
        DISC_RADIUS * IMAGE_SIZE
    )

    cv2.circle(
        disc_outline,
        disc_center,
        disc_radius,
        (0, 255, 0),
        2
    )

    # Display results
    plt.figure(
        figsize=(16, 4)
    )

    plt.subplot(
        1,
        4,
        1
    )

    plt.imshow(display_image)
    plt.title("Original image")
    plt.axis("off")

    plt.subplot(
        1,
        4,
        2
    )

    plt.imshow(heatmap, cmap="jet")
    plt.title("Grad-CAM")
    plt.axis("off")

    plt.subplot(
        1,
        4,
        3
    )

    plt.imshow(overlay)
    plt.title("Grad-CAM overlay")
    plt.axis("off")

    plt.subplot(
        1,
        4,
        4
    )

    plt.imshow(disc_outline)
    plt.title("Approximate disc region")
    plt.axis("off")

    plt.tight_layout()

    if save_output:

        safe_name = os.path.splitext(
            os.path.basename(image_path)
        )[0]

        figure_path = os.path.join(
            RELIABILITY_RESULTS_DIR,
            safe_name + "_explanation.png"
        )

        plt.savefig(
            figure_path,
            dpi=300,
            bbox_inches="tight"
        )

        print(
            "\nExplanation saved to:",
            figure_path
        )

    plt.show()

    result = {
        "filename": os.path.basename(image_path),
        "glaucoma_probability": glaucoma_probability,
        "normal_probability": 1 - glaucoma_probability,
        "uncertainty": uncertainty,
        "quality_score": quality_result["quality_score"],
        "attention_alignment": attention_alignment,
        "reliability_score": reliability_score,
        "decision": decision
    }

    return result

# ============================================================
# 14. Analyze one validation image
# ============================================================

example_image_path = validation_dataframe.iloc[0]["filepath"]

example_result = analyze_one_image(
    image_path=example_image_path,
    save_output=True
)

# ============================================================
# 15. Analyze all validation images
# ============================================================

print("\nAnalyzing all validation images...")

all_results = []

for index in range(
    len(validation_dataframe)
):

    image_path = validation_dataframe.iloc[
        index
    ]["filepath"]

    result = analyze_one_image(
        image_path=image_path,
        save_output=False
    )

    result["actual_label"] = int(
        validation_dataframe.iloc[index]["label"]
    )

    result["actual_class"] = (
        "glaucoma"
        if result["actual_label"] == 1
        else "normal"
    )

    all_results.append(result)

results_dataframe = pd.DataFrame(
    all_results
)

results_csv_path = os.path.join(
    RELIABILITY_RESULTS_DIR,
    "reliability_results.csv"
)

results_dataframe.to_csv(
    results_csv_path,
    index=False
)

print(
    "\nReliability results saved to:",
    results_csv_path
)

# ============================================================
# 16. Coverage and accepted-case analysis
# ============================================================

results_dataframe["accepted"] = (
    ~results_dataframe["decision"].str.startswith(
        "review required"
    )
)

accepted_dataframe = results_dataframe[
    results_dataframe["accepted"]
].copy()

if len(accepted_dataframe) > 0:

    accepted_dataframe["predicted_label"] = (
        accepted_dataframe[
            "glaucoma_probability"
        ] >= GLAUCOMA_THRESHOLD
    ).astype(int)

    accepted_accuracy = accuracy_score(
        accepted_dataframe["actual_label"],
        accepted_dataframe["predicted_label"]
    )

    print("\nSelective prediction results")
    print("==============================")
    print(
        "Total validation images:",
        len(results_dataframe)
    )

    print(
        "Accepted images:",
        len(accepted_dataframe)
    )

    print(
        "Rejected/review images:",
        len(results_dataframe) -
        len(accepted_dataframe)
    )

    print(
        "Coverage:",
        f"{len(accepted_dataframe) / len(results_dataframe):.4f}"
    )

    print(
        "Accuracy on accepted images:",
        f"{accepted_accuracy:.4f}"
    )

else:

    print(
        "\nNo images passed the reliability rules."
    )

# ============================================================
# 17. Save final summary
# ============================================================

summary_path = os.path.join(
    RELIABILITY_RESULTS_DIR,
    "reliability_summary.txt"
)

with open(summary_path, "w") as file:

    file.write(
        "Reliability-Aware Glaucoma CNN Summary\n"
    )

    file.write(
        "========================================\n"
    )

    file.write(
        f"MC passes: {MC_PASSES}\n"
    )

    file.write(
        f"Glaucoma threshold: {GLAUCOMA_THRESHOLD}\n"
    )

    file.write(
        f"Normal threshold: {NORMAL_THRESHOLD}\n"
    )

    file.write(
        f"Quality threshold: {QUALITY_THRESHOLD}\n"
    )

    file.write(
        f"Uncertainty threshold: "
        f"{uncertainty_threshold}\n"
    )

    file.write(
        f"Total validation images: "
        f"{len(results_dataframe)}\n"
    )

    file.write(
        f"Accepted images: "
        f"{len(accepted_dataframe)}\n"
    )

print(
    "\nCompleted reliability-aware analysis."
)
```

```python
# ============================================================
# ACRIMA RELIABILITY-AWARE DECISION LAYER
# BasePaperCNN + Image Quality + MC Dropout + Grad-CAM
# ============================================================

# ============================================================
# 1. IMPORTS
# ============================================================

import os
import cv2
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from PIL import Image
from sklearn.metrics import accuracy_score


# ============================================================
# 2. CONFIGURATION
# ============================================================

MC_PASSES = 30

# Prediction thresholds
GLAUCOMA_THRESHOLD = 0.90
NORMAL_THRESHOLD = 0.10

# MC-Dropout uncertainty calibration
UNCERTAINTY_PERCENTILE = 90

# Image-quality threshold
QUALITY_THRESHOLD = 0.60

# Grad-CAM alignment threshold
ALIGNMENT_THRESHOLD = 0.20

# Combined reliability threshold
RELIABILITY_THRESHOLD = 0.60

# Approximate optic-disc region
# IMPORTANT:
# This is only a prototype approximation.
DISC_CENTER_X = 0.50
DISC_CENTER_Y = 0.50
DISC_RADIUS = 0.22

# Model output type
# Your BasePaperCNN ends with Linear(..., 1),
# therefore it produces ONE LOGIT.
MODEL_OUTPUT_TYPE = "logit"

# Results directory
RELIABILITY_RESULTS_DIR = os.path.join(
    RESULTS_DIR,
    "reliability_results"
)

os.makedirs(RELIABILITY_RESULTS_DIR, exist_ok=True)


# ============================================================
# 3. CHECK REQUIRED VARIABLES
# ============================================================

required_variables = [
    "model",
    "DEVICE",
    "validation_transforms",
    "validation_dataframe",
    "RESULTS_DIR",
    "IMAGE_SIZE"
]

missing_variables = [
    variable
    for variable in required_variables
    if variable not in globals()
]

if missing_variables:
    raise RuntimeError(
        "Missing required variables: "
        + ", ".join(missing_variables)
        + "\n\nRun the BasePaperCNN training cell first."
    )


print("=" * 60)
print("RELIABILITY-AWARE ACRIMA ANALYSIS")
print("=" * 60)

print("\nReliability results will be saved to:")
print(RELIABILITY_RESULTS_DIR)

print("\nBase model:")
print(model)

print("\nDevice:")
print(DEVICE)

print("\nAssumed class mapping:")
print("Class 0 = Normal")
print("Class 1 = Glaucoma")


# ============================================================
# 4. MOVE MODEL TO DEVICE
# ============================================================

model = model.to(DEVICE)


# ============================================================
# 5. DISABLE IN-PLACE RELU
# ============================================================
# This prevents the Grad-CAM autograd/view conflict.
#
# It does NOT change learned weights.
# It only changes:
#
#     ReLU(inplace=True)
#
# into:
#
#     ReLU(inplace=False)
# ============================================================

def disable_inplace_relu(module):

    changed_count = 0

    for child in module.modules():

        if isinstance(child, nn.ReLU) and child.inplace:

            child.inplace = False
            changed_count += 1

    return changed_count


changed_relu_count = disable_inplace_relu(model)

print(
    f"\nChanged {changed_relu_count} ReLU layers "
    "from inplace=True to inplace=False."
)


# ============================================================
# 6. VERIFY MODEL OUTPUT FORMAT
# ============================================================

print("\nChecking model output format...")

model.eval()

with torch.no_grad():

    dummy_input = torch.randn(
        1,
        3,
        IMAGE_SIZE,
        IMAGE_SIZE
    ).to(DEVICE)

    dummy_output = model(dummy_input)

print("Model output shape:", tuple(dummy_output.shape))

if dummy_output.numel() == 1:

    print("Detected binary single-logit output.")

elif (
    dummy_output.ndim == 2
    and dummy_output.shape[1] == 2
):

    print("Detected two-class output.")

else:

    raise RuntimeError(
        f"Unsupported model output shape: {tuple(dummy_output.shape)}"
    )


# ============================================================
# 7. OUTPUT → GLAUCOMA PROBABILITY
# ============================================================
# Handles BOTH:
#
# Case 1:
#     Linear(..., 1)
#     sigmoid(logit)
#
# Case 2:
#     Linear(..., 2)
#     softmax(...)[glaucoma]
# ============================================================

def get_glaucoma_probability(output):

    if MODEL_OUTPUT_TYPE == "logit":

        # ----------------------------------------------------
        # SINGLE LOGIT MODEL
        # ----------------------------------------------------

        if output.numel() == 1:

            glaucoma_probability = torch.sigmoid(
                output.reshape(-1)[0]
            )

            return glaucoma_probability

        # ----------------------------------------------------
        # TWO-LOGIT MODEL
        # ----------------------------------------------------

        elif (
            output.ndim == 2
            and output.shape[1] == 2
        ):

            probabilities = torch.softmax(
                output,
                dim=1
            )

            # Class 1 = Glaucoma
            return probabilities[:, 1]

        else:

            raise RuntimeError(
                "Unsupported output shape: "
                + str(tuple(output.shape))
            )

    else:

        # If model already returns probabilities
        if output.numel() == 1:

            return output.reshape(-1)[0]

        elif (
            output.ndim == 2
            and output.shape[1] == 2
        ):

            return output[:, 1]

        else:

            raise RuntimeError(
                "Unsupported probability output shape: "
                + str(tuple(output.shape))
            )


# ============================================================
# 8. GLAUCOMA TARGET SCORE FOR GRAD-CAM
# ============================================================

def get_glaucoma_target_score(output):

    # --------------------------------------------------------
    # SINGLE LOGIT
    # --------------------------------------------------------

    if output.numel() == 1:

        return output.reshape(-1)[0]

    # --------------------------------------------------------
    # TWO LOGITS
    # --------------------------------------------------------

    elif (
        output.ndim == 2
        and output.shape[1] == 2
    ):

        # Class 1 = Glaucoma
        return output[:, 1].sum()

    else:

        raise RuntimeError(
            "Unsupported model output shape: "
            + str(tuple(output.shape))
        )


# ============================================================
# 9. IMAGE QUALITY FUNCTIONS
# ============================================================

def calculate_image_quality(image):

    """
    Returns a quality score between 0 and 1.

    Components:
        - Sharpness
        - Brightness
        - Contrast
        - Black-border penalty
    """

    image_array = np.array(image)

    if image_array.ndim == 3:

        gray = cv2.cvtColor(
            image_array,
            cv2.COLOR_RGB2GRAY
        )

    else:

        gray = image_array

    # --------------------------------------------------------
    # 1. Sharpness
    # --------------------------------------------------------

    laplacian_variance = cv2.Laplacian(
        gray,
        cv2.CV_64F
    ).var()

    sharpness_score = np.clip(
        laplacian_variance / 500.0,
        0.0,
        1.0
    )

    # --------------------------------------------------------
    # 2. Brightness
    # --------------------------------------------------------

    brightness = np.mean(gray) / 255.0

    brightness_score = 1.0 - min(
        abs(brightness - 0.50) / 0.50,
        1.0
    )

    # --------------------------------------------------------
    # 3. Contrast
    # --------------------------------------------------------

    contrast = np.std(gray) / 128.0

    contrast_score = np.clip(
        contrast,
        0.0,
        1.0
    )

    # --------------------------------------------------------
    # 4. Black border percentage
    # --------------------------------------------------------

    black_pixels = np.sum(gray < 10)

    total_pixels = gray.size

    black_ratio = (
        black_pixels /
        max(total_pixels, 1)
    )

    border_score = np.clip(
        1.0 - black_ratio * 2.0,
        0.0,
        1.0
    )

    # --------------------------------------------------------
    # Final quality score
    # --------------------------------------------------------

    quality_score = (

        0.35 * sharpness_score

        + 0.25 * brightness_score

        + 0.25 * contrast_score

        + 0.15 * border_score

    )

    return float(
        np.clip(
            quality_score,
            0.0,
            1.0
        )
    )


# ============================================================
# 10. ENABLE MC DROPOUT
# ============================================================

def enable_mc_dropout(model):

    """
    Keeps the model in evaluation mode,
    but activates Dropout layers.
    """

    model.eval()

    dropout_count = 0

    for module in model.modules():

        if isinstance(
            module,
            (
                nn.Dropout,
                nn.Dropout1d,
                nn.Dropout2d,
                nn.Dropout3d
            )
        ):

            module.train()
            dropout_count += 1

    return dropout_count


# ============================================================
# 11. MC DROPOUT PREDICTION
# ============================================================

def mc_dropout_prediction(
    image_tensor,
    model,
    device,
    passes=30
):

    """
    Performs multiple stochastic forward passes.

    Returns:
        mean probability
        variance
        standard deviation
        entropy
        all probabilities
    """

    dropout_count = enable_mc_dropout(model)

    if dropout_count == 0:

        print(
            "\nWARNING: No Dropout layers found."
            "\nMC Dropout uncertainty will not be meaningful."
        )

    probabilities = []

    image_tensor = image_tensor.to(device)

    with torch.no_grad():

        for _ in range(passes):

            output = model(image_tensor)

            glaucoma_probability = (
                get_glaucoma_probability(output)
                .reshape(-1)[0]
                .item()
            )

            probabilities.append(
                float(glaucoma_probability)
            )

    probabilities = np.array(
        probabilities,
        dtype=np.float32
    )

    mean_probability = float(
        np.mean(probabilities)
    )

    variance = float(
        np.var(probabilities)
    )

    standard_deviation = float(
        np.std(probabilities)
    )

    # --------------------------------------------------------
    # Predictive entropy
    # --------------------------------------------------------

    p = np.clip(
        mean_probability,
        1e-8,
        1.0 - 1e-8
    )

    entropy = float(
        -(
            p * np.log(p)
            +
            (1.0 - p) * np.log(1.0 - p)
        )
    )

    model.eval()

    return {

        "mean_probability":
            mean_probability,

        "variance":
            variance,

        "std":
            standard_deviation,

        "entropy":
            entropy,

        "all_probabilities":
            probabilities

    }


# ============================================================
# 12. GRAD-CAM
# ============================================================
#
# IMPORTANT FIX:
#
# OLD APPROACH:
#
#     register_full_backward_hook()
#
# This caused:
#
#     Output of BackwardHookFunctionBackward is a view
#     and is being modified inplace
#
# NEW APPROACH:
#
#     Forward hook
#     +
#     Tensor-level gradient hook
#
# This avoids the backward module-hook conflict.
# ============================================================

class GradCAM:

    def __init__(
        self,
        model,
        target_layer,
        device
    ):

        self.model = model
        self.target_layer = target_layer
        self.device = device

        self.activations = None
        self.gradients = None

        # Only a forward hook is used.
        self.forward_handle = (
            target_layer.register_forward_hook(
                self.forward_hook
            )
        )

    # --------------------------------------------------------
    # Forward hook
    # --------------------------------------------------------

    def forward_hook(
        self,
        module,
        input_data,
        output
    ):

        # Save activation values.
        self.activations = output.detach()

        # IMPORTANT:
        # Register gradient directly on the tensor.
        if output.requires_grad:

            output.register_hook(
                self.gradient_hook
            )

    # --------------------------------------------------------
    # Tensor gradient hook
    # --------------------------------------------------------

    def gradient_hook(self, gradient):

        self.gradients = gradient.detach()

    # --------------------------------------------------------
    # Generate CAM
    # --------------------------------------------------------

    def generate(
        self,
        image_tensor
    ):

        self.model.eval()

        self.model.zero_grad(
            set_to_none=True
        )

        self.activations = None
        self.gradients = None

        image_tensor = image_tensor.to(
            self.device
        )

        output = self.model(
            image_tensor
        )

        # Explain glaucoma score
        target_score = (
            get_glaucoma_target_score(
                output
            )
        )

        # Backpropagation
        target_score.backward()

        if self.activations is None:

            raise RuntimeError(
                "Grad-CAM activations were not captured."
            )

        if self.gradients is None:

            raise RuntimeError(
                "Grad-CAM gradients were not captured."
            )

        activations = self.activations
        gradients = self.gradients

        # ----------------------------------------------------
        # Global average pooling of gradients
        # ----------------------------------------------------

        weights = torch.mean(
            gradients,
            dim=(2, 3),
            keepdim=True
        )

        # ----------------------------------------------------
        # Weighted feature maps
        # ----------------------------------------------------

        cam = torch.sum(
            weights * activations,
            dim=1
        )

        cam = torch.relu(cam)

        # ----------------------------------------------------
        # Normalize
        # ----------------------------------------------------

        cam = cam[0].detach().cpu().numpy()

        cam_min = cam.min()
        cam_max = cam.max()

        if cam_max - cam_min > 1e-8:

            cam = (
                (cam - cam_min)
                /
                (cam_max - cam_min)
            )

        else:

            cam = np.zeros_like(cam)

        # ----------------------------------------------------
        # Resize to image size
        # ----------------------------------------------------

        cam = cv2.resize(
            cam,
            (
                IMAGE_SIZE,
                IMAGE_SIZE
            )
        )

        return cam

    # --------------------------------------------------------
    # Remove hook
    # --------------------------------------------------------

    def close(self):

        if self.forward_handle is not None:

            self.forward_handle.remove()

            self.forward_handle = None


# ============================================================
# 13. FIND GRAD-CAM TARGET LAYER
# ============================================================

print("\nSelecting Grad-CAM target layer...")

if not hasattr(model, "features"):

    raise RuntimeError(
        "BasePaperCNN does not contain a 'features' module."
    )

if len(model.features) <= 6:

    raise RuntimeError(
        "model.features[6] does not exist."
    )

target_layer = model.features[6]

print("Grad-CAM target layer:")
print(target_layer)

gradcam = GradCAM(
    model=model,
    target_layer=target_layer,
    device=DEVICE
)

print("Grad-CAM initialized successfully.")


# ============================================================
# 14. APPROXIMATE OPTIC-DISC MASK
# ============================================================

def create_disc_mask(
    height,
    width
):

    mask = np.zeros(
        (height, width),
        dtype=np.uint8
    )

    center_x = int(
        DISC_CENTER_X * width
    )

    center_y = int(
        DISC_CENTER_Y * height
    )

    radius = int(
        DISC_RADIUS *
        min(height, width)
    )

    cv2.circle(
        mask,
        (
            center_x,
            center_y
        ),
        radius,
        1,
        -1
    )

    return mask


# ============================================================
# 15. GRAD-CAM ATTENTION ALIGNMENT
# ============================================================

def calculate_attention_alignment(
    cam,
    disc_mask
):

    # --------------------------------------------------------
    # Find high-attention region
    # --------------------------------------------------------

    threshold = np.percentile(
        cam,
        80
    )

    high_attention = (
        cam >= threshold
    )

    # --------------------------------------------------------
    # Alignment
    # --------------------------------------------------------

    total_attention_pixels = (
        np.sum(high_attention)
    )

    if total_attention_pixels == 0:

        return 0.0

    aligned_pixels = np.sum(
        high_attention &
        (disc_mask > 0)
    )

    alignment = (
        aligned_pixels /
        total_attention_pixels
    )

    return float(
        np.clip(
            alignment,
            0.0,
            1.0
        )
    )


# ============================================================
# 16. RELIABILITY SCORE
# ============================================================

def calculate_reliability_score(
    quality_score,
    uncertainty,
    attention_alignment,
    uncertainty_threshold
):

    # --------------------------------------------------------
    # Normalize uncertainty relative to calibrated threshold.
    #
    # This is better than dividing by a fixed 0.25 because
    # MC-dropout variance is usually much smaller.
    # --------------------------------------------------------

    normalized_uncertainty = np.clip(
        uncertainty /
        max(
            uncertainty_threshold,
            1e-8
        ),
        0.0,
        1.0
    )

    uncertainty_reliability = (
        1.0 -
        normalized_uncertainty
    )

    # --------------------------------------------------------
    # Weighted reliability score
    # --------------------------------------------------------

    reliability = (

        0.40 * quality_score

        + 0.35 * uncertainty_reliability

        + 0.25 * attention_alignment

    )

    return float(
        np.clip(
            reliability,
            0.0,
            1.0
        )
    )


# ============================================================
# 17. FINAL DECISION
# ============================================================

def make_final_decision(
    glaucoma_probability,
    quality_score,
    uncertainty,
    attention_alignment,
    uncertainty_threshold
):

    # --------------------------------------------------------
    # Reliability score
    # --------------------------------------------------------

    reliability_score = (
        calculate_reliability_score(
            quality_score=quality_score,
            uncertainty=uncertainty,
            attention_alignment=attention_alignment,
            uncertainty_threshold=uncertainty_threshold
        )
    )

    # --------------------------------------------------------
    # Step 1: Image quality
    # --------------------------------------------------------

    if quality_score < QUALITY_THRESHOLD:

        decision = "REVIEW_REQUIRED"
        reason = "Poor image quality"

    # --------------------------------------------------------
    # Step 2: MC Dropout uncertainty
    # --------------------------------------------------------

    elif uncertainty > uncertainty_threshold:

        decision = "REVIEW_REQUIRED"
        reason = "High prediction uncertainty"

    # --------------------------------------------------------
    # Step 3: Grad-CAM alignment
    # --------------------------------------------------------

    elif attention_alignment < ALIGNMENT_THRESHOLD:

        decision = "REVIEW_REQUIRED"
        reason = "Low attention alignment"

    # --------------------------------------------------------
    # Step 4: Combined reliability
    # --------------------------------------------------------

    elif reliability_score < RELIABILITY_THRESHOLD:

        decision = "REVIEW_REQUIRED"
        reason = "Low overall reliability"

    # --------------------------------------------------------
    # Step 5: High-confidence glaucoma
    # --------------------------------------------------------

    elif glaucoma_probability >= GLAUCOMA_THRESHOLD:

        decision = "GLAUCOMA"
        reason = "High-confidence reliable prediction"

    # --------------------------------------------------------
    # Step 6: High-confidence normal
    # --------------------------------------------------------

    elif glaucoma_probability <= NORMAL_THRESHOLD:

        decision = "NORMAL"
        reason = "High-confidence reliable prediction"

    # --------------------------------------------------------
    # Step 7: Borderline probability
    # --------------------------------------------------------

    else:

        decision = "REVIEW_REQUIRED"
        reason = "Borderline prediction probability"

    return (
        decision,
        reason,
        reliability_score
    )


# ============================================================
# 18. CALIBRATE UNCERTAINTY THRESHOLD
# ============================================================

print("\n")
print("=" * 40)
print("CALIBRATING UNCERTAINTY THRESHOLD")
print("=" * 40)

calibration_uncertainties = []

for index, row in validation_dataframe.iterrows():

    image_path = row["filepath"]

    try:

        image = Image.open(
            image_path
        ).convert("RGB")

        image_tensor = (
            validation_transforms(image)
            .unsqueeze(0)
        )

        mc_result = (
            mc_dropout_prediction(
                image_tensor=image_tensor,
                model=model,
                device=DEVICE,
                passes=MC_PASSES
            )
        )

        calibration_uncertainties.append(
            mc_result["variance"]
        )

    except Exception as error:

        print(
            f"Skipping {image_path}: {error}"
        )


if len(calibration_uncertainties) == 0:

    raise RuntimeError(
        "No validation images were successfully "
        "processed during uncertainty calibration."
    )


UNCERTAINTY_THRESHOLD = float(
    np.percentile(
        calibration_uncertainties,
        UNCERTAINTY_PERCENTILE
    )
)

print(
    "\nCalibrated uncertainty threshold:",
    UNCERTAINTY_THRESHOLD
)

print(
    f"Using {UNCERTAINTY_PERCENTILE}th percentile "
    "of validation uncertainty."
)


# ============================================================
# 19. ANALYZE ONE IMAGE
# ============================================================

def analyze_one_image(
    image_path,
    true_label=None,
    save_output=False,
    show_output=True
):

    # --------------------------------------------------------
    # Load image
    # --------------------------------------------------------

    image = Image.open(
        image_path
    ).convert("RGB")

    # --------------------------------------------------------
    # Image quality
    # --------------------------------------------------------

    quality_score = (
        calculate_image_quality(
            image
        )
    )

    # --------------------------------------------------------
    # Transform image
    # --------------------------------------------------------

    image_tensor = (
        validation_transforms(image)
        .unsqueeze(0)
        .to(DEVICE)
    )

    # --------------------------------------------------------
    # MC Dropout
    # --------------------------------------------------------

    mc_result = (
        mc_dropout_prediction(
            image_tensor=image_tensor,
            model=model,
            device=DEVICE,
            passes=MC_PASSES
        )
    )

    glaucoma_probability = (
        mc_result["mean_probability"]
    )

    uncertainty = (
        mc_result["variance"]
    )

    # --------------------------------------------------------
    # Grad-CAM
    # --------------------------------------------------------

    cam = gradcam.generate(
        image_tensor
    )

    # --------------------------------------------------------
    # Approximate optic disc mask
    # --------------------------------------------------------

    disc_mask = create_disc_mask(
        IMAGE_SIZE,
        IMAGE_SIZE
    )

    # --------------------------------------------------------
    # Attention alignment
    # --------------------------------------------------------

    attention_alignment = (
        calculate_attention_alignment(
            cam=cam,
            disc_mask=disc_mask
        )
    )

    # --------------------------------------------------------
    # Final decision
    # --------------------------------------------------------

    (
        decision,
        reason,
        reliability_score
    ) = make_final_decision(

        glaucoma_probability=
            glaucoma_probability,

        quality_score=
            quality_score,

        uncertainty=
            uncertainty,

        attention_alignment=
            attention_alignment,

        uncertainty_threshold=
            UNCERTAINTY_THRESHOLD
    )

    # --------------------------------------------------------
    # True label
    # --------------------------------------------------------

    true_label_name = None

    if true_label is not None:

        true_label_name = (
            "Glaucoma"
            if int(true_label) == 1
            else "Normal"
        )

    # --------------------------------------------------------
    # Visualization
    # --------------------------------------------------------

    if show_output or save_output:

        display_image = np.array(
            image.resize(
                (
                    IMAGE_SIZE,
                    IMAGE_SIZE
                )
            )
        )

        heatmap = np.uint8(
            255 * cam
        )

        heatmap = cv2.applyColorMap(
            heatmap,
            cv2.COLORMAP_JET
        )

        heatmap = cv2.cvtColor(
            heatmap,
            cv2.COLOR_BGR2RGB
        )

        overlay = cv2.addWeighted(
            display_image,
            0.60,
            heatmap,
            0.40,
            0
        )

        # ----------------------------------------------------
        # Disc mask visualization
        # ----------------------------------------------------

        disc_visualization = (
            display_image.copy()
        )

        disc_visualization[
            disc_mask > 0
        ] = (
            0.5 *
            disc_visualization[
                disc_mask > 0
            ]
            + 0.5 *
            np.array(
                [255, 255, 255]
            )
        ).astype(np.uint8)

        # ----------------------------------------------------
        # Create figure
        # ----------------------------------------------------

        fig, axes = plt.subplots(
            1,
            4,
            figsize=(18, 5)
        )

        axes[0].imshow(
            display_image
        )

        axes[0].set_title(
            "Original Image"
        )

        axes[1].imshow(
            cam,
            cmap="jet"
        )

        axes[1].set_title(
            "Grad-CAM"
        )

        axes[2].imshow(
            overlay
        )

        axes[2].set_title(
            "Grad-CAM Overlay"
        )

        axes[3].imshow(
            disc_visualization
        )

        axes[3].set_title(
            "Approx. Disc Region"
        )

        for ax in axes:

            ax.axis("off")

        plt.suptitle(
            f"Decision: {decision}\n"
            f"Probability: {glaucoma_probability:.3f} | "
            f"Quality: {quality_score:.3f} | "
            f"Uncertainty: {uncertainty:.5f} | "
            f"Alignment: {attention_alignment:.3f} | "
            f"Reliability: {reliability_score:.3f}",
            fontsize=12
        )

        plt.tight_layout()

        if save_output:

            filename = os.path.splitext(
                os.path.basename(image_path)
            )[0]

            output_path = os.path.join(
                RELIABILITY_RESULTS_DIR,
                f"{filename}_gradcam.png"
            )

            plt.savefig(
                output_path,
                dpi=150,
                bbox_inches="tight"
            )

            print(
                "\nGrad-CAM saved to:",
                output_path
            )

        if show_output:

            plt.show()

        else:

            plt.close(fig)

    # --------------------------------------------------------
    # Return result
    # --------------------------------------------------------

    return {

        "filepath":
            image_path,

        "true_label":
            true_label,

        "true_label_name":
            true_label_name,

        "glaucoma_probability":
            glaucoma_probability,

        "normal_probability":
            1.0 - glaucoma_probability,

        "uncertainty":
            uncertainty,

        "uncertainty_std":
            mc_result["std"],

        "predictive_entropy":
            mc_result["entropy"],

        "image_quality":
            quality_score,

        "attention_alignment":
            attention_alignment,

        "reliability_score":
            reliability_score,

        "decision":
            decision,

        "reason":
            reason

    }


# ============================================================
# 20. TEST ONE VALIDATION IMAGE
# ============================================================

print("\n")
print("=" * 40)
print("TESTING ONE VALIDATION IMAGE")
print("=" * 40)

first_row = validation_dataframe.iloc[0]

test_result = analyze_one_image(

    image_path=
        first_row["filepath"],

    true_label=
        first_row["label"],

    save_output=True,

    show_output=True
)

print("\nPrediction Report")
print("-" * 40)

print(
    "True label:",
    test_result["true_label_name"]
)

print(
    "Glaucoma probability:",
    f"{test_result['glaucoma_probability']:.4f}"
)

print(
    "Normal probability:",
    f"{test_result['normal_probability']:.4f}"
)

print(
    "MC uncertainty:",
    f"{test_result['uncertainty']:.6f}"
)

print(
    "Image quality:",
    f"{test_result['image_quality']:.4f}"
)

print(
    "Attention alignment:",
    f"{test_result['attention_alignment']:.4f}"
)

print(
    "Reliability score:",
    f"{test_result['reliability_score']:.4f}"
)

print(
    "Final decision:",
    test_result["decision"]
)

print(
    "Reason:",
    test_result["reason"]
)


# ============================================================
# 21. ANALYZE ALL VALIDATION IMAGES
# ============================================================

print("\n")
print("=" * 40)
print("ANALYZING ALL VALIDATION IMAGES")
print("=" * 40)

all_results = []

for index, row in validation_dataframe.iterrows():

    image_path = row["filepath"]

    true_label = row["label"]

    try:

        result = analyze_one_image(

            image_path=image_path,

            true_label=true_label,

            save_output=False,

            show_output=False
        )

        all_results.append(result)

        if (
            len(all_results) % 10 == 0
        ):

            print(
                f"Processed "
                f"{len(all_results)}/"
                f"{len(validation_dataframe)}"
            )

    except Exception as error:

        print(
            f"\nError processing:"
            f"\n{image_path}"
            f"\n{error}"
        )


# ============================================================
# 22. SAVE RESULTS
# ============================================================

results_dataframe = pd.DataFrame(
    all_results
)

csv_path = os.path.join(
    RELIABILITY_RESULTS_DIR,
    "reliability_validation_results.csv"
)

results_dataframe.to_csv(
    csv_path,
    index=False
)

print(
    "\nReliability results saved to:"
)

print(csv_path)


# ============================================================
# 23. SELECTIVE PREDICTION METRICS
# ============================================================

if len(results_dataframe) > 0:

    accepted_mask = (
        results_dataframe["decision"].isin(
            [
                "GLAUCOMA",
                "NORMAL"
            ]
        )
    )

    accepted_results = (
        results_dataframe[
            accepted_mask
        ]
    )

    total_images = len(
        results_dataframe
    )

    accepted_images = len(
        accepted_results
    )

    coverage = (
        accepted_images /
        max(total_images, 1)
    )

    print("\n")
    print("=" * 40)
    print("SELECTIVE PREDICTION RESULTS")
    print("=" * 40)

    print(
        f"Total images: {total_images}"
    )

    print(
        f"Accepted images: {accepted_images}"
    )

    print(
        f"Rejected / referred images: "
        f"{total_images - accepted_images}"
    )

    print(
        f"Coverage: {coverage:.4f}"
    )

    # --------------------------------------------------------
    # Accepted accuracy
    # --------------------------------------------------------

    if accepted_images > 0:

        y_true = (
            accepted_results[
                "true_label"
            ].astype(int)
        )

        y_pred = (
            accepted_results[
                "decision"
            ]
            .map(
                {
                    "NORMAL": 0,
                    "GLAUCOMA": 1
                }
            )
            .astype(int)
        )

        accepted_accuracy = (
            accuracy_score(
                y_true,
                y_pred
            )
        )

        print(
            f"Accepted prediction accuracy: "
            f"{accepted_accuracy:.4f}"
        )

    else:

        print(
            "No predictions were accepted."
        )

    # --------------------------------------------------------
    # Decision distribution
    # --------------------------------------------------------

    print("\nDecision distribution:")

    print(
        results_dataframe[
            "decision"
        ].value_counts()
    )


# ============================================================
# 24. SAVE SUMMARY
# ============================================================

summary_path = os.path.join(
    RELIABILITY_RESULTS_DIR,
    "reliability_summary.txt"
)

with open(
    summary_path,
    "w"
) as file:

    file.write(
        "ACRIMA Reliability-Aware Analysis\n"
    )

    file.write(
        "=================================\n\n"
    )

    file.write(
        "Base model: BasePaperCNN\n"
    )

    file.write(
        "Class 0: Normal\n"
    )

    file.write(
        "Class 1: Glaucoma\n\n"
    )

    file.write(
        f"MC passes: {MC_PASSES}\n"
    )

    file.write(
        f"Uncertainty percentile: "
        f"{UNCERTAINTY_PERCENTILE}\n"
    )

    file.write(
        f"Uncertainty threshold: "
        f"{UNCERTAINTY_THRESHOLD}\n"
    )

    file.write(
        f"Quality threshold: "
        f"{QUALITY_THRESHOLD}\n"
    )

    file.write(
        f"Alignment threshold: "
        f"{ALIGNMENT_THRESHOLD}\n"
    )

    file.write(
        f"Reliability threshold: "
        f"{RELIABILITY_THRESHOLD}\n"
    )

    file.write(
        f"Glaucoma probability threshold: "
        f"{GLAUCOMA_THRESHOLD}\n"
    )

    file.write(
        f"Normal probability threshold: "
        f"{NORMAL_THRESHOLD}\n"
    )

    file.write(
        "\nReliability score weights:\n"
    )

    file.write(
        "Image quality = 0.40\n"
    )

    file.write(
        "MC uncertainty = 0.35\n"
    )

    file.write(
        "Grad-CAM alignment = 0.25\n"
    )

print(
    "\nSummary saved to:"
)

print(summary_path)


# ============================================================
# 25. CLOSE GRAD-CAM HOOK
# ============================================================

gradcam.close()

print("\n")
print("=" * 60)
print("RELIABILITY ANALYSIS COMPLETED")
print("=" * 60)
```
